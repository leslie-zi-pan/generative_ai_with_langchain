# Tokenization

Lets try some tokenisation example from transformers library

```python
from transformers import AutoTokenizer

# Load a pretrained tokeniser
model_name = "meta-llama/Meta-Llama-3-8B" # deepseek-ai/DeepSeek-V2
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenise a text
text = "Kyrie Irving is the best basketball player in the world"

encoded_text = tokenizer(text)

# Access te tokenized IDs
print(tokenizer.decode(encoded_text["input_ids"]))

print(encoded_text["input_ids"])
print(", ".join([tokenizer.decode(t) for t in encoded_text["input_ids"]]))
```
